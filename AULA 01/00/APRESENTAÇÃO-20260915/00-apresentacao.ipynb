## Objectivos
---

- Aplicar técnicas adequadas para melhorar a qualidade de um conjunto de dados
- Extrair informação a partir de um conjunto de dados, e visualizar essa informação de forma efetiva

## Programa
---

- Introdução às ferramentas informáticas para análise e visualização de dados
- Analise de Dados
    - Obtenção de dados; Limpeza e preparação de dados; Integração, agrupamento e transformação de dados; Exploração dos dados
- Visualização da Informação
    - Princípios básicos; Tabelas e gráficos; Visualização de informação univariada e multivariada; Criação de painéis (_dashboards_) de informação

## Avaliação
---

- Componentes:
    - Trabalho(s) Prático(s) (TP)
    - Provas Escritas (PE): Testes / Exames
- Componentes classificadas de 0-20 valores, com classificação mínima de 7 valores
- A nota final é dada pela média ponderada

$$\text{Nota Final} = TP \times 0.5 + PE \times 0.5$$

- O aluno fica aprovado se a $\text{Nota Final} \geq 9.5$
