Crie um `DataFrame` (`df4`) com os seguintes dados:

|---------|---------|--------------|

1. Converta a coluna **Salario** para o tipo `float`.
2. Crie uma nova coluna **Bonus**, onde o valor é $5\%$ do salário de cada funcionário.
3. Calcule o salário total (salário + bónus) e adicione-o numa nova coluna.

```python
# Calcular salario + bonus e colocar em nova coluna

df4['Total'] = df4['Salario'] + df4['Bonus']
df4
```

```python
# Acessar somente aos funcionarios que estao em TI e marketing

df4[(df4['Departamento'] == 'Marketing') | (df4['Departamento'] == 'TI')]
```

```python
# Selecionar todos os funcionarios com o salario acima de 3000

df4[df4['Salario'] > 3000]
```

```python
# Criar novo dataframe chamado  df_bempagos contendo funcionarios que ganhem mais de 3800

df_bempagos = df4[df4['Total'] > 3800]
df_bempagos = pd.DataFrame(df_bempagos)
df_bempagos
```

## Exercício 7

1. Crie uma `Series` com os valores `[200, 500, 250, 100]`, onde o índice representa produtos: `['Produto A', 'Produto B', 'Produto C', 'Produto D']`.
2. Crie um tipo de categorias ordenadas, com os valores "Barato" e "Caro" e "Luxo".
3. Crie outra `Series`, resultado de aplique uma função à primeira `Series` que categorize os produtos como "Barato" se o valor for menor que 180, "Luxo" se for 500 ou mais, e "Caro" todos os outros. Não se esqueça que o tipo deve ser o criado na alínea anterior.
4. Crie um `DataFrame` concatenando as duas `Series`, e altere o nome das colunas para "Preço" e "Categoria".
5. Aplique uma segunda função ao `DataFrame` para atualizar o "Preço", de forma a efetuar um desconto de 10% para da categoria "Caros" ou superior.
6. Calcule a média dos Preços.
7. Calculo o produto dos Preços.
8. Determine quantos produtos têm Preço superior a 150.

```python
# Crie uma Series com os valores [200, 500, 250, 100], onde o índice representa produtos: ['Prodtudo A', 'Prodtuo B',
# 'Produto C', 'Produto D'].

s7 = pd.Series([200, 500, 250, 100], index=['Produto A', 'Produto B', 'Produto C', 'Produto D'])
s7
```

```python
# Crie um tipo de categorias oredenadas, com os valores "Barato e "Caro" e "Luxo"

qualidade_cat = pd.CategoricalDtype(['Barato','Caro','Luxo'], ordered=True)
qualidade_cat
```

```python
# Crie outra Series, resultado de aplique uma função á primeira Series que categorize os produtos como "Barato" se o 
# valor form menor que 180, "Luxo" se for 500 ou mais e "Caro" todos os outros. 
# Não se esqueça que o tipo dese ser criado na alinea anterior.

def categoriza(valor):
    if valor < 180:
        return "Barato"

    elif valor >= 500:
        return "Luxo"

    else:
        return "Caro"

s7new = s7.apply(categoriza)
s7new = s7new.astype(qualidade_cat)

s7new
```

```python
# Crie um DataFrame concatenando as duas Series, e altere o nome para "Preço" e "Categoria"

df7 = pd.concat([s7, s7new], axis=1)
df7.columns = ['Preço','Categoria']
df7
```

```python
# Aplique uma segunda função ao seu DataFrame para atualizar um "Preço", 
# de forma a efetuar um desconto de 10% para da categoria "Caros" ou superior.

def desconto(row):
    if row['Categoria'] >= 'Caro':
        return row ['Preço'] * 0.9
    else:
        return row ['Preço']

df7['Preço'] = df7.apply(desconto, axis=1)
df7
```

```python
# Calcular media dos preços

df7['Preço'].mean()
```

```python
# Calcule o produto dos preços

produto = df7['Preço'].prod()
produto
```

```python
# Determine quantos produtos tem o preço superior a 150

len(df7[df7['Preço'] > 150])
```
