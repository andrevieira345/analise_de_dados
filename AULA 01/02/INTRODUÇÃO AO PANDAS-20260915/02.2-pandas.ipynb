# Introdução ao `pandas`

O pandas é a livraria _quasi-standart_ para trabalhar com dados, utiliazando o Python, pois implementa praticamente todas as funções necessárias para trabalhar, analisar e manipular os dados. No site do [pandas](https://pandas.pydata.org/) está a seguinte frase, que resume o conceito da livraria:

> _pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool,
built on top of the Python programming language_.

Antes de se utilizar esta livraria, é necessário fazer a sua importação. A forma normal de importar o pandas é a fazer a importação com o alias `pd`.

````python
import pandas as pd

pd.options.display.max_columns = 999
````

O `pandas` possui duas estruturas básicas para armazenar a informação: `Series` e `DataFrame`.

## `Series`

No `pandas`, a estrutura `Series` é uma estrutura unidimensional, e o seu conteúdo, apesar de poder ser de tipos diferentes, é normalmente sempre do mesmo tipo.

A forma mais elementar de criar `Series`, é passar uma lista com os valores de cada elemento:

````python
nums = pd.Series([0, 0.25, 0.5, 0.75, 2])
nums
````

Pode-se ver que, ao contrário do que foi dito anteriormente, a `Series` parece que é bidimensional, mas é apenas a forma da estrutura ser apresentada, pois para além da sequência de valores, contém também um índice inteiro, entre `0` e `N-1` (onde `N` corresponde ao número de elementos). Pode-se aceder a cada uma destas sequências utilizando os atributos `values` e `index`.

````python
nums.values
````

````python
nums.index
````

E pode-se aceder aos elementos utilizando a forma normal do Python, através de parênteses retos, apesar de exigir alguma atenção, como veremos adiante.

````python
nums[2]
````

````python
nums[1:3]
````

````python
nums[4] = 1
nums
````

Mas ao contrário das listas do Python ou dos `arrays` do Numpy, cujos indices são sempre numéricos, nas `Series` os  índices podem ser definidos, passando uma lista com os índices no argumento `index`.

````python
songs = pd.Series([145, 142, 38, 13], index=['Paul', 'John', 'George', 'Ringo'], name='counts')
songs
````

E claro que agora podemos utilizar esses índices para aceder aos elementos:

````python
songs['Paul']
````

E eventualmente, adicionar novas linhas aos dados:

````python
songs['Anonymous'] = 0
songs
````

### `Series` como dicionários especializados

De certa forma, pode-se pensar numa `Series` como se fosse uma forma especializada de um dicionário do Python. Enquanto um dicionário faz o mapeamento entre chaves e valores arbitrários, uma `Series` é uma estrutura que faz o mapeamento entre chaves de um determinado tipo e valores de um determinado tipo. A definição do tipo é importante, pois é isso que permite uma maior eficiência desta estrutura em certas operações.

A analogia entre `Series` e dicionários pode ser ainda mais evidente se se considerar que se pode criar diretamente uma `Series` a partir de um dicionário, sendo as chaves os índices da `Series`.

````python
population_dict = {
    'Nova Iorque': 20201249,
    'Florida': 21538187,
    'Pensilvania': 13002700,
    'California': 39538223,
    'Texas': 29145505,
}
````

````python
population = pd.Series(population_dict, name='population')
population
````

Quando se utiliza um dicionário como base, a ordem dos elementos será de acordo com a ordem das chaves no dicionário, que por sua vez depende da ordem de inserção no mesmo. Isso pode ser definido utilizando o parametro `index`, passando quais as chaves que devem ser utilizadas, e em que ordem devem aparecer.

````python
states_list = ['California', 'Florida', 'Ohio', 'Texas']
pop2 = pd.Series(population_dict, index=states_list)
pop2
````

Três coisas podem-se observar neste exemplo:

1. as chaves que existiam no dicionário, mas que não foram indicadas no `index`, não aparecem na `Series`;
2. as chaves que foram indicadas no `index`, mas não não existem no dicionário vão conter o valor `NaN`;
3. por outro lado, enquanto no primeiro exemplo o tipo de dados foi `int64` porque todos os valores eram inteiros, neste último exemplo, devido à necessidade do pandas utilizar o `NaN`, o tipo é `float64`, pois este tipo suporta `NaN` enquanto `int64` não suporta.

`NaN` (_Not a Number_) é um valor definido no numpy para indicar valores em falta, ou `NA`, ou `None`, sendo que o pandas ignora estes valores em muitas das suas operações. Por exemplo, vamos comparar o método `.count` que conta os valores existentes na `Series`, com a propriedade `.size`, que contém o numero de entradas (incluindo valores em falta):

````python
pop2.size
````

````python
pop2.count()
````

Uma `Series` pode ser transformada num dicionário utilizando o método `.to_dict()`.

````python
pop2.to_dict()
````

Ou numa lista com o método `.to_list()`, sendo que neste caso apenas os valores serão devolvidos.

````python
population.to_list()
````

````python
population.values
````

### Categorias

Quando se sabe que a diversidade dos dados está limitada a apenas uns quantos valores diferentes, pode-se utilizar categorias, que têm alguns beneficios:

* Utilizam menos memória do que _strings_
* Tem maior performance
* Podem ser ordenadas
* Pode-se realizar operações nas categorias
* Impor a pertença aos valores

Para criar uma categoria é preciso passar o argumento `dtype='category'` no construtor da `Series`, ou então utilizar o método `astype` a uma `Series` já criada, passando o valor `'category'`.

````python
sizes_list = ['XS', 'M', 'XL', 'L', 'S']
sizes = pd.Series(sizes_list, dtype='category')
sizes
````

Como se pode ver, a `Series` é uma categoria, com 5 elementos.

Se a categoria representa um tamanho, a ordem pode ser importante, mas por defeito as categorias não têm nenhuma ordem, como se pode ser observando o atributo `.cat`, que contêm algumas propriedades.

````python
sizes.cat.*?
````

````python
sizes.cat.ordered
````

A forma mais simples de criar categorias ordenadas, em vez de passar o argumento `'category'` na criação da `Series`, é definir um novo tipo, utilizando o tipo `CategoricalDtype`, com `ordered=True`.

````python
sizes_cat = pd.CategoricalDtype(['S', 'M', 'L'], ordered=True)
sizes2 = pd.Series(sizes_list, dtype=sizes_cat)
sizes2
````

Quando as categorias têm uma ordem, pode-se efetuar comparações.

````python
sizes2 >= 'M'
````

Outra forma de criar uma ordem nas categorias é utilizar o método `.reorder_categories()` a uma categoria existente. Mas atenção que desta forma é necessário indicar todos os membros da categoria, caso contrário irá ocorrer um `ValueError`.

````python
sizes3 = sizes.cat.reorder_categories(['XS', 'S', 'M', 'L', 'XL'], ordered=True)
sizes3
````

````python
sizes3 >= 'L'
````

## `DataFrame`

O `DataFrame` é a outra estrutura básica do pandas, e pode ser vista como uma tabela rectangular de dados, contendo uma coleção ordenada de colunas com nomes, onde cada coluna pode conter valores de diferentes tipos.

Se uma `Series` é análoga a uma estrutura uni-dimensional com índices explícitos, um `DataFrame` é análogo a uma estrutura bidimensional, com índices para as linhas e para as colunas explícitos.

Um `DataFrame` pode ser vista como uma sequencia de `Series` alinhadas, i.e., todas as `Series` partilham o mesmo índice.

<div class="alert alert-block alert-info">
<p>ℹ️<b>Nota</b></p><p>Olhar para um <code>DataFrame</code> como um conjunto de linhas vai parecer "esquisito", pois a sintaxe de utilização não é muito coerente com essa forma de pensar. Será mais correto pensar num <code>DataFrame</code> como um conjunto de colunas, onde cada coluna é uma <code>Series</code></p>
</div>

Existem várias formas de criar um `DataFrame`, uma das mais comuns é utilizar um dicionário de listas do mesmo tamanho.

````python
data = {
    'nome': ['Ricardo', 'Ana Rita', 'Patricia', 'José', 'Inês', 'João'],
    'alt': [1.78, 1.67, 1.87, 1.74, 1.79, 1.9],
    'num': [1234, 7634, 3433, 4234, 6543, 4321]
}
````

O `DataFrame` resultante terá o índice automático (como nas `Series`), e as colunas aparecem na ordem das chaves no dicionário.

````python
df = pd.DataFrame(data)
print(df)
df
````

Caso de indique uma lista de colunas, elas irão aparecer nessa ordem, e caso o nome indicado não exista no dicionário, irá aparecer como valores em falta.

````python
alunos = pd.DataFrame(data, columns=['num', 'nome', 'alt', 'media'])
alunos
````

````python
func = pd.DataFrame(data2)
func
````

Ou utilizando `Series` previamente definidas.

````python
capital_dict = {'California': 'Sacramento',
                'Florida': 'Tallahassee',
                'Nova Iorque': 'Albany',
                'Pensilvania': 'Harrisburg',
                'Texas': 'Austin',
               }
````

O exemplo acima parece "esquisito"... Isso porque por defeito, a função `concat` faz a concatenação utilizando o eixo (_axis_) 0 (linhas), portanto vai concatenar um objecto e depois o outro "em baixo, de acordo com os índices das colunas.

Outra coisa que o exemplo acima serve para mostrar é que o _label_ dos índices não são "chaves primárias", ou seja não têm a obrigatoriedade de serem únicos, como se pode ver acima, onde o nome de cada estado aparece duas vezes.

### `.loc` e `.iloc`

Para aceder às linhas, pode-se utilizar os atributos especiais `.loc` e `.iloc`. A diferença entre estes dois atributos é que enquanto o primeiro (`.loc`) utiliza o _nome_ (_labels_) dos índices/colunas, o segundo (`.iloc`) utiliza uma notação _posicional_ (similar aos aos elementos de uma lista).

Estes atributos especiais são válidos quer nas `Series` quer no `DataFrame`. No caso de uma `Series`, devolve o valor correspondente, no caso do `DataFrame`, pode devolve uma `Series` ou um valor, dependendo do que for indicado.

````python
nomes.loc[1234]          # o index é o 1234
````

````python
nomes.iloc[0]           # é a "posição" 0 (a primeira)
````

No caso de um `DataFrame`, caso seja indicado apenas um índice ("linha"), devolve um `Series` (uma "linha" do `DataFrame).

````python
alunos
````

````python
s = alunos.loc[1234]
s
````

````python
type(s)
````

````python
alunos.iloc[2]
````

Também se pode indicar dois índices ("linha" e "coluna"), e nesse caso será devolvido um valor, que corresponde ao elemento correspondente. A linha e a coluna pode ser indicada de duas formas: Ou uma notação _recursiva_, (`df.loc[linha][coluna]`) (mais comum no Python) ou uma notação mais _matricial_, através de uma lista de índices separada por uma virgula (`df.loc[linha, coluna]`).

````python
alunos
````

````python
alunos.loc[1234, 'nome']
````

````python
alunos.loc[1234]['nome']
````

````python
alunos.iloc[4, 2]
````

````python
alunos.iloc[4].iloc[2]
````

O erro acima está relacionado com a forma como o pandas interpreta a sintaxe `[][]`. Primeiro será realizado o `alunos.iloc[4]`, que devolve um `Series` e depois vai fazer `Series[2]`, sem utilizar o `iloc` o que vai provocar o `FutureWarning`, como já vimos anteriormente.

<div class="alert alert-block alert-info">
<p><b>Nota</b></p>
   <p>Um erro comum a quem está a começar é utilizar <code>.loc</code> e <code>.iloc</code> como se fossem métodos (utilizando parênteses), quanto na realidade eles são atributos.</p>
</div>

### _Slices_

Pode-se utilizar a sintaxe de _slices_, para os índices, mas cuidado, o funcionamento é diferente caso se utilize _labels_ ou _posições_.

Quando se utiliza _labels_, o índice final **é incluído** na _slice_, ao contrário do que acontece com as listas no Python, mas se se utilizar _posições_ funciona da forma habitual (o elemento final **não é incluído**)

````python
alunos.loc[1234:3433]
````

````python
alunos.iloc[0:2]
````

Pode-se utilizar _slices_ nos dois índices:

````python
alunos.loc[1234:3433, 'nome':'curso']
````

````python
alunos.iloc[0:2, 0:2]
````

Quando se quer aceder a alguns elementos de um eixo, mas a todos os elementos do outro eixo, pode-se utilizar o sinal de dois ponto (`:`) no eixo onde se quer todos os elementos:

````python
alunos.loc[:, 'nome']
````

No outro eixo pode-se indicar uma _slice_ para obter um `DataFrame`:

````python
alunos.loc[:, 'nome':'curso']
````

Mas também se pode indicar uma lista de índices a selecionar:

````python
alunos.loc[:, ['nome', 'curso']]
````

````python
alunos.iloc[:, [0, 2]]
````

Tal como nas `Series`, pode-se utilizar _arrays_ booleanos para filtrar valores de um `DataFrame`:

````python
alunos[alunos.alt > 1.75]
````

Que também podem ser utilizados com o `.loc`:

````python
alunos.loc[alunos.alt > 1.75]
````

E a partir da versão >3.0 também funcionam com o `.iloc`, apesar de não ser recomendado a sua utilização.

````python
alunos.iloc[alunos.alt > 1.75]
````

Como se viu, existem muitas forma de selecionar e rearranjar os dados no pandas. Abaixo apresenta-se um breve sumário de algumas delas.

|<div style="width:180px">Sintaxe</div> | Comentário |
|---------------------------------------|------------|
`df[colunas]`              | Seleciona uma coluna ou subconjunto de colunas do `DataFrame`; alguns casos especiais são o uso de _arrays_ booleanos (filtra linhas) ou _slices posicionais_ (filtra linhas)
`df.loc[linhas]`           | Seleciona uma linha ou um subconjunto de linhas do `DataFrame`, utilizando a(s) _label(s)_ indicada
`df.loc[:, colunas]`      | Seleciona uma coluna ou um subconjunto de colunas utilizando a(s) _label(s)_ indicadas
`df.loc[linhas, colunas]` | Seleciona a(s) linha(s) e coluna(s) indicadas pela(s) respetiva(s) _label(s)_
`df.iloc[linhas]`         | Seleciona uma linha ou um subconjunto de linhas indicadas pela respetiva(s) _posição(ões)_
`df.iloc[:, colunas]`     | Seleciona uma coluna ou um sobconjunto de colunas utilizando a(s) _posição(ões)_ indicadas
`df.iloc[linhas, colunas]`| Seleciona a(s) linha(s) e coluna(s) indicadas pela(s) respetiva(s) _posição(ões)_

## Remover entradas

Para remover entradas utiliza-se o método `.drop`, com a ressalva que, por defeito, o método não remove do objecto indicado, mas sim cria e devolve uma cópia do objecto, com as respetivas entradas removidas (praticamente todos os métodos do pandas trabalham desta forma). No caso das `Series`pode-se simplesmente indicar o índice da entrada, ou uma lista de entradas a remover:

````python
nomes = alunos['nome']
nomes
````

````python
nomes.drop(1234)
````

````python
nomes.drop([4321, 1234, 6543])
````

Também se pode utilizar os parâmetros `labels` ou `index` (são sinónimos).

````python
nomes.drop(labels=[1234,4321])
````

````python
nomes
````

Nenhuma das operações anteriores alterou o objecto base (`nomes` no caso), foi sempre devolvida uma cópia com as alterações pretendidas. Caso se pretenda alterar o objecto deve-se utilizar o parâmetro `inplace=True`.

````python
nomes.drop(7634, inplace=True)
nomes
````

Quando aplicado a um `DataFrame`, o método `drop` por defeito irá remover as entradas das "linhas".

````python
alunos.drop(1234)
````

Pode-se indicar a(s) linha(s) e/ou coluna(s) a remover utilizando respetivamente os parâmetros `index` e `columns`.

````python
alunos.drop(index=[1234, 4321, 6543], columns='curso')
````

Também se pode utilizar uma lista de índices, e o parâmatro `axis`, indicado `0` ou `index` (para as linhas) ou `1` ou `columns` (para as colunas).

````python
alunos.drop(['alt','curso'], axis=1)    # ou axis='columns'
````

````python
alunos.drop([1234, 4321], axis='index')
````

## Aritmética e alinhamento das entradas

É muito fácil trabalhar com objectos que tenham índices diferentes, pois quando nesse caso o pandas fará a _união_ dos objectos.

````python
s1 = pd.Series(range(1,5), index=list('abcd'))
s1
````

````python
s2 = pd.Series(range(10, 6, -1), index=list('bdfa'))
s2
````

Adicionando estas `Series`, iremos obter:

````python
s1 + s2
````

Como se pode ver, o alinhamento interno das entrada irá inserir valores em falta nos índices que não sejam comuns, e esses valores em falta serão propagados para as operações aritméticas seguintes.

O mesmo acontece com os `DataFrame`:

````python
# Criar duas "matrizes" com valores aleatórios
import random
m1 = [[random.randint(1,10) for _ in range(4)] for _ in range(3)]
m2 = [[random.randint(1,10) for _ in range(3)] for _ in range(4)]
````

````python
df1 = pd.DataFrame(m1, columns=list('abcd'))
df1
````

````python
df2 = pd.DataFrame(m2, columns=list('bde'))
df2
````

````python
df1 + df2
````

Existem métodos nos objetos do panda equivalentes aos operadores aritméticos. Por exemplo, o método `add` (equivalente ao operador `+`) permite somar dois `DataFrame` (ou duas `Series`):

````python
df1.add(df2)
````

A vantagem da utilização destes métodos é que frequentemente o que se pretende é utilizar um valor por defeito (por exemplo o `0`), caso existem valores em falta. Isso pode ser feito passando o argumento `fill_value` com o valor pretendido:

````python
df1.add(df2, fill_value=0)
````

Para cada método aritmético, existe um semelhante que começa com a letra `r`, em que os argumentos são revertidos. Por exemplo:

````python
df1.sub(df2, fill_value=0)      # df1 - df2
````

````python
df2.rsub(df1, fill_value=0)     # df1 - df2
````

Abaixo uma tabela com os métodos aritméticos:

| Método | Descrição |
|--------|-----------|
|`add, radd`           | Métodos para adição (`+`) |
|`sub, rsub`           | Métodos para subtração (`-`) |
|`div, rdiv`           | Métodos para divisão (`/`) |
|`floordiv, rfloordiv` | Métodos para divisão inteira (`//`) |

## Utilização e mapeamento de funções

É frequente querer-se aplicar uma função uni-dimensional a cada linha/coluna. O método `apply` faz exatamente isso. Por exemplo, vamos calcular a amplitude:

````python
def amplitude(x):
    return x.max() - x.min()
````

````python
df1
````

````python
df1.apply(amplitude)
````

Podia-se ter obtido o mesmo resultado utilizando uma função anónima:

````python
df1.apply(lambda x: x.max()-x.min())
````

O resultado é uma `Series`, resultado da aplicação da função a cada coluna do `DataFrame`. O índice da `Series` são as colunas do `DataFrame`.

Caso se pretenda fazer ao contrário (aplicar a função a cada linha do `DataFrame`), deve-se utilizar o parâmetro `axis`, indicado `1` ou `columns`.

````python
df1.apply(amplitude, axis=1)
````

As funções a utilizar não estão limitadas à devolução de um único valor, podem devolver multiplos valores:

````python
def minmax(x):
    return pd.Series([x.min(), x.max()], index=['min', 'max'])
````

````python
df1.apply(minmax, axis=1)
````

O `DataFrame` implementa métodos que calculam muitas funções estatísticas nesses casos não é necessário utilizar o método `apply`.

````python
df1.apply(sum)     # A função sum (do Python) é o argumento do método apply do DataFrame
````

````python
df1.sum()          # Aplicar diretamente o método sum do DataFrame
````

Caso o objectivo seja aplicar uma função elemento-a-elemento pode-se utilizar o método `map`:

````python
df1.map(lambda x: f'{x:.2f}')
````

````python
s1.map(lambda x: 5*x)
````

## Ordenação

Por vezes é necessário/útil poder ordenar quer os índices quer os valores. Isso pode ser feito através dos métodos `sort_index` e `sort_values`.

````python
alunos
````

Por defeito, o método `sort_index` ordena o `index` (linhas):

````python
alunos.sort_index()
````

Mas, utilizando o parâmetro `axis` pode-se indicar qual o eixo (linhas/colunas) a utilizar.

````python
alunos.sort_index(axis=1, ascending=False)
````

No exemplo anterior também se pode ser que, por defeito, é feita uma ordenação ascendente, mas utilizando o parâmetro `ascending` pode-se indicar que se pretende uma ordenação descendente.

O método `sort_values` pode ser utilizado diretamente numa `Series`:

````python
s2.sort_values()
````

Mas no caso de um `DataFrame`, é necessário indicar qual/quais a(s) coluna(s) que deve(m) ser utilizada(s) para ordenar os dados. Pode-se passar a _label_ da coluna, ou uma lista de _labels_, ou utilizando o parâmetro `by`.

````python
alunos.sort_values(by='curso')
````

Caso se pretenda ordenar utilizando diferentes subcritérios (diversas colunas), pode-se indicar para cada critério se pretende ordenar de forma ascendente ou descendente.

````python
alunos.sort_values(by=['curso', 'nome'], ascending=False)
````

Pode-se também indicar se se quer que os valores em falta (`NaN`) apareçam no final (`last`, valor por defeito) ou no inicio (`first`) da listagem, utilizando o parâmetro `na_position`:

````python
alunos.sort_values(by='curso', na_position='first')
````

Relembrar que estes método, por defeito, não alteram o objecto sob o qual estão a operar, mas devolvem um cópia do objecto com as alterações. Caso se queira alterar diretamente o objecto deve-se utilizar o parâmetro `inplace=True`.

````python
alunos
````

## Métodos estatísticos

Os objectos do pandas possuem um conjunto de métodos que implementam algumas das funções matemáticas e estatísticas mais comuns, de forma computacionalmente eficiente.

Como se viu anteriormente, pode-se somar as entradas (por defeito percorre as linhas) de um objecto utilizando o método `sum`:

````python
df1 = pd.DataFrame([[5.2, pd.NA],
                    [6.1, 3.2],
                    [pd.NA, pd.NA],
                    [4.5, 3.4]],
                   index=list('abcd'),
                   columns=list('xy'))
df1
````

````python
df1.sum()
````

Caso se pretenda a soma percorrendo as colunas, deve-se utilizar o parâmetro `axis`, indicando `1` 
ou `columns`:

````python
df1.sum(axis='columns')  # axis=1
````

Por defeito, caso existam valores em falta estes serão ignorados, caso todas as entradas da linha/coluna sejam `NA`, então a soma será `0`. Passar o parâmetro `skipna=False` faz com que seja devolvido o valor `NA` caso alguma das entradas seja `NA`.

````python
df1.sum(axis=1, skipna=False)
````

Alguns dos métodos exigem que pelo um dos valores não seja `NA` para devolver um resultado, um desses métodos é o `mean`:

````python
df1.mean(axis='columns')
````

Outros métodos devolvem estatísticas indiretas, como os métodos `idxmin` e `idxmax`, que devolvem os índices onde se encontra o valor mínimo/máximo.

````python
df1
````

````python
df1.idxmax()
````

Existem ainda métodos que são _comulativos_:

````python
df1
````

````python
df1.cumsum()
````
