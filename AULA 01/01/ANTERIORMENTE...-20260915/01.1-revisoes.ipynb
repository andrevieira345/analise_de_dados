### Operadores aritméticos

````python
m = 10
n = 3
````

````python
print(f"{m ** n = }")
print(f"{m * n  = }")
print(f"{m / n  = }")
print(f"{m // n = }")
print(f"{m % n  = }")
print(f"{m + n  = }")
print(f"{m - n  = }")
````

Nos exemplos acima pode-se ver uma característica das _f-strings_, que é particularmente útil quando se está a testar código, que é a auto-documentação das expressões. Para utilizar basta escrever um sinal de igual (`=`) no final da expressão que se quer avaliar, e o _output_ será a expressão, o sinal de igual e o resultado da avaliação da expressão. Os espaços antes ou depois do sinal de igual não são importantes, podem existir ou não, mas serão reflectidos no _output_.

### Operadores de comparação

Permitem comparar dois (ou mais) valores, e o resultado será sempre um valor lógico

* `<` Menor do que
* `>` Maior do que
* `<=` Menor ou igual do que
* `>=` Maior ou igual do que
* `==` Igual a
* `!=` Diferente de

````python
x = 10
y = 20
z = 30
w = 10
````

````python
print(f"{x < y = }")
print(f"{x > z = }")
print(f"{y <= z = }")
print(f"{x >= y = }")
print(f"{x == w = }")
print(f"{y != z = }")
````

> **NOTA:** O _output_ da última linha não aparece igual aos outros porque se está a utilizar o ponto de exclamação (`!`) dentro da expressão da _f-string_, e esse caractere é utilizado da especificação das mesmas (ver a documentação em [https://docs.python.org/pt-br/3/reference/lexical_analysis.html#f-strings]())
>
>Na prática isto praticamente nunca seria um problema, por duas razões: a primeira é porque esta utilização apenas deve ser utilizada para efetuar o _debugging_/testes, e a segunda, o que deveríamos fazer era definir uma variável com o resultado, e depois apresentar essa variável. Por exemplo:
>
> ```python
> resultado = y != z
> print(f"{resultado = }")
> ```

````python
print(f"{x < y < z = }")
print(f"{x < w < z = }")
````

No caso das comparações com mais do que dois valores (**comparações encadeadas**), na prática o que acontece internamente pelo interpretador de Python é efetuar as comparações dos diferentes blocos, e depois "juntar" o resultado dos diferentes blocos.

Por exemplo, considere-se a seguinte comparação:

```python
4 < 5 > 3
```

Internamente serão avaliadas individualmente as comparações `4 < 5` e `5 > 3`, e caso ambas sejam verdadeiras, o resultado será verdadeiro, caso alguma delas seja falsa, o resultado também o será.

````python
4 < 5 > 3
````

````python
4 < 5 > 3 < 1
````

### `list`

As listas são a estrutura de dados mais fundamental do Python. Basicamente é uma coleção ordenada de dados, que podem ou não ser do mesmo tipo. As listas são **mutáveis**, i.e., pode-se alterar o conteúdo da lista depois de criada, e são definidas através de parênteses retos (`[`) ou da função `list` (para converter outra estrutura numa lista).

````python
uma_lista = [1, True, 5.3, "Ola", None]
print(uma_lista)
````

````python
outra_lista = list("Olá, como vais?")
print(outra_lista)
````

````python
len(uma_lista)
````

````python
len(outra_lista)
````

Depois da lista criada, pode-se aceder ao n-ésimo elemento de uma lista utilizando parênteses retos (`[`)

````python
x = [10, 11, 12, 13, 14, 15]
print(x[0])
print(x[1])
print(x[-1])
x[0] = -1
print(x)
````

Os parênteses retos também servem para indicar **fatias** (_slices_) de uma lista. A fatia `i:j` significa que são os elementos desde o `i` (inclusive) até ao `j` (exclusive). Caso não seja indicado o inicio da fatia será considerado o inicio da lista, e caso não seja indicado o fim da fatia, será considerado o fim da lista.

````python
x = [10, 11, 12, 13, 14, 15, 16]
print(x[1:3])
print(x[:3])
print(x[4:])
print(x[-2:])
````

A fatia pode ainda conter um terceiro elemento, que definirá o passo.

````python
print(x[2::3])
print(x[::-1])
````

### `tuple`

Tuplos são uma estrutura quase idêntica às listas, com uma diferença fundamental: Os tuplos são **imutáveis**, ou seja, um tuplo, depois de criado não pode ser alterado. Tuplos são definidos utilizando parênteses (`(`), ou a função `tuple`, para converter uma outra estrutura num tuplo.

Os tuplos são de tal forma similares às listas, que tudo o que foi visto acima para as listas é válido par aos tuplos, **excepto** o que altera o conteúdo da lista, visto que os tuplos não podem ser alterados.

````python
um_tuplo = (1.2, 4)
um_tuplo
````

````python
outro_tuplo = tuple(x)
outro_tuplo
````

Tal como as listas, pode-se aceder a qualquer elemento de um tuplo indicando o índice do mesmo, entre parênteses retos (`[ ]`). Mas como os tuplos são **imutáveis**, tentar alterar um tuplo depois de criado vai provocar um erro.

````python
print(um_tuplo[0])
````

````python
um_tuplo[0] = 2
````

Os tuplos (e as listas também) podem ser _descompactados_, i.e., podem-se definir uma variável para cada elemento do tuplo de forma semi-automática.

````python
a, b = um_tuplo
print(a)
print(b)
````

### `dict`

Outra estrutura fundamental são os dicionários, que permitem guardar um conjunto de pares _chave_-_valor_, associando cada _chave_ ao respetivo _valor_. Dessa forma pode-se aceder, alterar, modificar ou apagar o respetivo valor. Em cada dicionário, cada chave deve ser única, e deve ser **imutável**, ou seja, uma lista não pode ser um chave de um dicionário, mas um tuplo pode.

Para criar um dicionários pode-se utilizar chavetas (`{ ... }`), ou a função `dict`.

````python
vazio = {}
vazio
````

### `set`

Finalmente temos os conjuntos (`set`), que de forma resumida, são uma estrutura de dados não ordenada de elementos únicos (sem repetição). Os conjuntos podem ser definidos pela função `set`, ou utilizando chavetas (`{...}`).

````python
s1 = set([1, 2, 3, 1, 4, 2, 3])
s1
````

Os conjuntos suportam operações sobre conjuntos, como a união, intersecção, diferença e diferença simétrica. Existem métodos e operadores que implementam estas operações, devolvendo o respetivo resultado

* Intersecção: `intersection` ou `&`
* Diferença: `difference` ou `-`
* Diferença simétrica: `symetric_diference` ou `^`

````python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7, 8}
````

````python
a.union(b)
````

````python
a | b
````

````python
a.intersection(b)             # a & b
````

````python
a.difference(b)               # a - b
````

````python
a.symmetric_difference(b)     # a ^ b
````

### `if`

A estrutura `if` é uma estrutura de decisão, que avalia uma expressão, e **se** o resultado dessa expressão seja verdadeiro, executa o bloco de código indicado

````python
n = -1

if n < 0:
    n = 0
    print("negativo passou a zero\n")
elif n == 0:
    print("nulo\n")
elif 0 <= n <= 9:
    print("um unico digito\n")
else:
    print("vários digitos\n")
````

Podem existir zero os várias partes `elif` (`elif` é a abreviatura de _else if_), e o parte do `else` é opcional.

### `while`

A estrutura `while` é uma estrutura de repetição, que vai avaliar uma expressão, e **enquanto** essa expressão for avaliada como verdadeira, o bloco de código indicado será executado.

````python
n = 2

while n > 0:
    print(n)
    n -= 1
````

### `for`

A estrutura `for` também é uma estrutura de repetição, mas vai iterar sobre os elementos de uma sequência (mais sobre sequência posteriormente), na ordem com que aparecem na sequência.

````python
nomes = ["Monty", "Python", "rulez"]
for nome in nomes:
    print(nome, end="!\n")
````

````python
for i in range(len(nomes)):
    print(nomes[i], end="!\n")
````

### `break` e `continue`

Os comandos `break` e `continue` permitem alterar o fluxo de execução dentro de uma estrutura de repetição (`while` ou `for`)

O comando `break` **termina imediatamente** a estrutura (salta para a próxima instrução **fora** da estrutura), e o comando `continue` **salta** para a próxima iteração da repetição.

````python
i = 0
while True:
    i += 1
    if i == 10:
        break
    elif 4 <= i <= 7:
        continue
    print(i)
````

### Algumas funções muito utilizadas nas estruturas de repetição

### `range(...)`

A função `range` vai criar uma progressão aritmética até ao limite fornecido (mas excluindo esse valor). Pode-se, opcionalmente, indicar o valor inicial e o passo.

````python
# lista de 0 a 5 (exclusive)
lista1 = range(5)
for i in lista1:
    print(i)
````

````python
# lista de 2 e 5 (exclusive)
lista2 = range(2, 5)
for i in lista2:
    print(i)
````

````python
# lista de 10 a 25 (exclusive), de 4 em 4
lista3 = range(10, 25, 4)
for i in lista3:
    print(i)
````

### `enumerate(...)`

Muitas vezes é necessário iterar sobre os elementos de uma sequência, mas também é necessário saber os índices desses mesmos elementos. Isso pode ser feito com a função `enumerate`, que recebe como argumento uma sequência, para cada elemento da sequência, devolve um tuplo com o índice e com o elemento.

````python
nomes = ["Monty", "Python", "rulez"]
````

````python
for i, nome in enumerate(nomes):
    print(i, nome)
````

````python
list(enumerate(nomes))
````

Sem a função `enumerate`, teríamos que recorrer às funções `range` e `len`, para criar uma progressão aritmética com tantos elementos quantos os elementos da sequência, e depois utilizar o índice para aceder a cada elemento.

````python
for i in range(len(nomes)):
    print(i, nomes[i])
````

## Funções

As funções são a principal e mais importante forma de organizar e reutilizar código. Tipicamente, caso se preveja a necessidade de repetir o mesmo código, ou muito semelhante, mais do que uma vez, provavelmente será uma boa ideia escrever uma função reutilizável. As funções também ajudam a tornar o código mais legível pois permitem atribuir um nome a um grupo de instruções.

### `def`

Em Python, as funções são definidas pela _keyword_ `def`, seguido do nome da função e, entre parênteses, os parâmetros de entrada. Dentro da função, pode existir ou não o comando `return`, que termina a função, devolvendo o valor indicado.

````python
def positivo(x):
    if x > 0:
        return True
    return False
````

````python
positivo(10)
````

````python
positivo(-45.2)
````

As funções podem ter parâmetros _opcionais_, que possuem um valor que será utilizado por defeito, caso não seja fornecido.

````python
def potencia(x, y=2):
    return x ** y
````

````python
potencia(4, 2)
````

````python
potencia(4)
````

````python
potencia(5, 5)
````

Também se podem indicar os _nomes_ dos parâmetros quando se chama a função, e dessa forma a ordem não é importante. Mas atenção que os argumentos _posicionais_ (aqueles que não se indica o nome do parâmetro) têm obrigatoriamente de vir em primeiro lugar.

````python
def volume(raio, altura=1):
    pi = 3.1415926
    return (raio*pi**2)*altura
````

````python
volume(4, 5)
````

````python
volume(altura=5, raio=4)
````

````python
volume(3, altura=5)
````

### Variáveis locais/globais

Não esquecer que as variáveis definidas e utilizadas _dentro_ da função são **variáveis locais**, ou seja, apenas são estão definidas enquanto a função está a ser executada, quando a função terminar essas variáveis deixam de existir

````python
def somar(x, y):
    soma = x + y
    return soma
````

````python
a = somar(1, 2)
````

````python
print(f"{a = }")
print(f"{x = }")
print(f"{y = }")
print(f"{soma = }")
````

Caso se queira alterar uma variável externa à função (uma **variável global**) deve-se definir essa variável dentro da função como `global`.

````python
def soma(a):
    global x
    x = 1
    x += a
````

````python
x = 10
soma(5)
print(x)
````

````python
x
````

### `lambda` (funções anónimas)

Por vezes é útil poder-se definir uma função anónima, não devido à sua reutilização, mas porque outra determinada função que queremos utilizar necessita de uma função como argumento, mas essa função não existe. As funções anónimas definem-se através da _keyword_ `lambda`.

Funções anónimas são muito utilizadas na análise dados, pois muitas funções de transformação de dados recebem funções como um dos seus argumentos.

````python
strings = ["foo", "bar", "card", "aaaa", "abab"]
````

````python
strings.sort()        # Ordena alfabeticamente
print(strings)
````

````python
strings.sort(key=len)  # Ordena pelo comprimento da string, utilizando a função len(...)
print(strings)
````

Caso se queira ordenar a lista de _strings_ pelo numero de letras repetidas em cada _string_, isso pode ser feito com uma função anónima.

````python
strings.sort(key=lambda x: len(set(x)))
print(strings)
````

Outro exemplo é ordenar um dicionário pelos valores.

````python
d = {4: "Tiago", 2:"André", 7: "Maria", 5:"Sonia"}
````

````python
print(sorted(d.keys()))       # Apenas as chaves
print(sorted(d.values()))     # Apenas os valores
print(sorted(d.items()))      # Por defeito, ordena pelas chaves
````

````python
print(sorted(d.items(), key=lambda item: item[1]))
````

Existem um sem número de funcionalidades do Python que não estão disponíveis por defeito, algumas vêm incluídas com o Python, outras são externas que têm de ser descarregadas e instaladas posteriormente. Para utilizar estas funcionalidades num programa, é necessário **importar** os módulos que as contêm.

Entre muitos outros, alguns dos módulos mais conhecidos e que fazem parte do Python são
* `csv`
* `datetime`
* `json`

Para importar os módulo para um programa, podemos usar o comando `import x`, ou o comando `from x import y`. Utilizando a primeira forma todas as funcionalidades do módulo `x` serão importadas, enquanto na segunda forma podemos indicar somente as funcionalidades que queremos importar.

Outra diferença fundamental é que enquanto o comando `import` exige que se utilize o _namespace_ do módulo antes da função, ou seja, depois do comando `import re`, todas as chamadas às funções do módulo `re` tinham de ser precedidas de `re.` para se poderem aceder. Isso não se passa com o `from x import y`, pois como de identifica quais os objectos a importar (`y` no caso), eles podem ser acedidos diretamente, sem a necessidade de indicar o _namespace_.

````python
import random
random.randint(1, 10)
````

````python
from pathlib import Path
print(Path().absolute())
````

O que não devem nunca fazer é `from x import *`. Pode parecer uma boa ideia, pois importa-se todo o módulo e não é necessário escrever constantemente o _namespace_, mas pode provocar erros posteriormente, pois pode-se sobrescrever acidentalmente algumas funções.

````python
from math import *        # Existe uma função dist no módulo math

dist([0,0], [1,1])        # Distância entre os pontos (0,0) e (1,1)
````

````python
dist = 20

dist([0,0], [1,1])
````

## Ficheiros

É importante perceber como funcionam as operações de leitura/escrita com ficheiros. É necessário _abrir_, _ler_/_escrever_ e depois _fechar_ o ficheiro.

Para abrir o ficheiro utiliza-se a função `open`, passando o caminho do ficheiro, o modo de abertura (leitura/escrita, texto/binário, etc.), entre outros parâmetros possíveis.

````python
f = open('01-revisoes.ipynb', 'r')
````

Depois _trabalha-se_ o ficheiro, quer lendo o seu conteúdo, quer escrevendo os dados pretendidos para o ficheiro. No exemplo abaixo o valor 500 indica que se pretende ler os unicamente os primeiros 500 caracteres (até ao fim da linha).

````python
for linha in f.readlines(100):
    print(linha.strip('\n'))        # remover o '\n' que é lido do ficheiro
````

Finalmente, fecha-se o ficheiro, quer para evitar corromper o ficheiro, quer para permitir abrir mais ficheiros (qualquer sistema operativo tem um limite do numero de ficheiro abertos em simultâneo)

````python
f.close()
````

A forma mais simples de trabalhar com ficheiros é utilizando o comando _with_, que permite criar um bloco que irá fechar o ficheiro de forma automática no término do bloco.

````python
with open('01-revisoes.ipynb', 'r', encoding='utf-8') as f:
    for linha in f.readlines(500):
        print(linha.strip('\n'))
````

## Exceções

Como vimos anteriormente, por vezes acontecem erros (exceções no contexto do Python), que podem ser controladas, caso seja esse o objectivo (e seja possível) de forma _graciosa_, de forma a tornar o programa mais robusto.

Isso é feito com as _keywords_ `try`/`except`.

````python
float('123.45')
````

````python
float([1,2])
````

````python
texto = "ola"
try:
    print(float(texto))
except ValueError:
    print("Não é possivel converter")
````

A clausula do `except`, apesar de poder ser utilizada sem mais nenhuma indicação, pode e **deve** ser sempre definida a exceção que estamos a tentar controlar, desse forma não se corre o risco de passar por cima de um _bug_ no programa. Caso seja necessário podem ser definidas várias clausulas `except`.

````python
texto = 'ola'
try:
    print(float(texto))
except ValueError:
    print("Não é possivel converter")
````

## Classes

Classes o conceito que serve de base à implementação em Python da **Programação Orientada a Objectos**.

Basicamente, e de forma muito resumida, um **objecto** é uma entidade que contêm dados e código. Os dados são as _propriedades_ (ou atributos) do objecto, e o código são os métodos, ou seja, as _acções_ que o objecto pode realizar.

As classes propriamente ditas são o **modelo** (em inglês é normalmente utilizado o termo _blueprint_) que define o objecto, enquanto o objecto é uma **instância** da classe.

````python
# Vamos definir um modelo para a classe individuo
class individuo:
    def __init__(self, nome):
        self.nome = nome
````

````python
# Vamos criar uma instância (um objecto) da classe individuo
pessoa = individuo("Pedro")
print(pessoa)
````

````python
class individuo:
    def __init__(self, nome):
        self.nome = nome

    def __str__(self):
        return self.nome
````

````python
pessoa = individuo("Pedro")
print(pessoa)
````

Através da **herança** podemos criar classes que têm como base outras classes previamente definidas, e que irão _herdar_ todas as características da classe base.

````python
class empregado(individuo):
    def __init__(self, nome, salario):
        super().__init__(nome)            # chamar o "construtor" da classe base
        self.salario = salario

    def __str__(self):
        return f"{self.nome} -> {self.salario}€"
````

````python
ana = empregado("Ana", 3500)
print(ana)
````
