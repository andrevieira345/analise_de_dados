# Exercícios de Revisões

Peça uma _string_ ao utilizador e imprima quantas letras maiúsculas estão presentes na _string_.

Escreva um programa que peça o nome de um ficheiro ao utilizador e indique qual tipo de programa que é capaz de ler esse ficheiro, analisando a sua extensão. O seu programa deve ser capaz de reconhecer pelo menos 5 tipos diferentes de ficheiros. Por exemplo, `"Cachorros.doc"` é um Documento do Word.

Escreva um programa para ser usado pela associação de estudantes na próxima eleição para calcular os votos após a eleição. O seu programa deve pedir o número de candidatos e os seus nomes. De seguida, o utilizador irá inserir todos os votos (1 para o primeiro candidato, 2 para o segundo candidato, ...) até ser inserido um 0. O programa informará quem venceu a eleição e qual percentagem obtida por cada candidato.

```
Quantos candidatos: 3
Candidato 1: Graham Chapman
Candidato 2: John Cleese
Candidato 3: Terry Gilliam

Indique o voto (0 para terminar): 2
Indique o voto (0 para terminar): 2
Indique o voto (0 para terminar): 2
Indique o voto (0 para terminar): 1

O John Cleese venceu
John Cleese obteve 75\% dos votos
Graham Chapman obteve 25\% dos votos
Terry Gilliam obteve 0\% dos votos
```

Escreva uma função que devolva os primeiros _n_ números da sequencia de Fibonacci ($1, 1, 2, 3, 5, 8, 13, 21, ...$)

Um palíndromo é uma palavra que se lê da mesma forma tanto da esquerda para a direita quanto da direita para a esquerda. Crie uma função chamada palindromo que recebe uma única palavra como parâmetro e devolva `True` se a palavra for um palíndromo e `False` caso contrário.

Crie as classes indicados abaixo, e instancie 1 ou 2 objectos de cada classe e teste as funcionalidades implementadas:

* Crie uma classe que permita definir um rectangulo. Quando instancia o objeto deve fornecer o comprimento e a largura do mesmo, e deve implementar métodos para devolver a área e o perimetro do mesmo.

* Crie outra classe, cuja classe base será o rectangulo, que implemente a definição de um triangulo rectangulo. Os dados para instanciar o objecto serão os mesmos (comprimento e largura), e deve fazer o _overload_ dos métodos para calcular a area e o perimetro.
