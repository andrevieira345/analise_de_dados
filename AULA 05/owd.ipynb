## 4. Uma necessidade concreta de `pivot`

CO2 tem uma linha por país–ano; energia tem uma por país–ano–indicador. Para combinar duas tabelas com uma linha por país–ano, transformamos os indicadores de energia em colunas.

```python
energia = energia.reset_index()
energia.head()
```

Confirmamos a chave

Usamos código ISO e ano como chaves, não o nome do país. Uma junção externa com `indicator=True` torna visíveis as observações exclusivas de cada fonte.


```python
juncao = co2.merge(
    energia,
    on=['iso_code', 'year'],
    how='outer',
    indicator=True,
)
juncao
```

O argumento `indicator` (acima) cria uma coluna (`_merge`, ou o nome indicado no argumento), com a origem dos dados da linha

```python
juncao['_merge'].value_counts()
```

Inspecionamos os casos presentes apenas na energia. O ano adicional explica a diferença de cobertura nos anexos, mas esta verificação é preferível a assumir que as fontes terminam no mesmo ano.

```python
juncao.loc[
    juncao['_merge'] == 'right_only',
    ['iso_code', 'year', 'country_y']
].head()
```

Sem sufixos explícitos, pandas distinguiu as colunas comuns com `_x` e `_y`. Vamos identificar esses nomes.

```python
juncao.columns.str.endswith(('_x', '_y'))
```

```python
#[c for c in juncao.columns if c.endswith(('_x', '_y'))]

juncao.columns[juncao.columns.str.endswith(('_x', '_y'))]
```

Refazemos a junção com sufixos informativos. As chaves aparecem uma vez; os nomes e a população aparecem nas duas versões.

```python
dados = co2.merge(
    energia,
    on=['iso_code', 'year'],
    how='inner',
    suffixes=('_co2', '_energia')
)
dados
```

Antes de eliminar colunas repetidas, verificamos os nomes e as populações. Um mesmo nome de variável não garante valores iguais nas duas fontes.

```python
comparacao_pop = dados['population_co2'] - dados['population_energia']
(comparacao_pop != 0).sum()
```

```python
(dados['country_co2'] != dados['country_energia']).sum()
```

Não usamos a diferença de população para concluir que uma fonte está errada: os CSV não documentam aqui a causa.

Neste caso, como são todos iguais, podemos remover uma das colunas e renomear de acordo

```python
dados = (
    dados
    .drop(['country_energia', 'population_energia'], axis=1)
    .rename({'country_co2': 'country', 'population_co2': 'population'}, axis=1)
)
dados
```

## 8. Uma Series hierárquica: `unstack` e `stack`
Selecionamos uma medida. O seu índice continua a ter país e ano.

```python
serie_co2 = painel['CO2']
serie_co2.head()
```

Passamos o país do índice para as colunas. Cada linha passa a representar um ano.

```python
co2_por_ano = serie_co2.unstack(['iso_code', 'country'])
co2_por_ano.head()
```

```python
co2_por_ano.loc[2020:2022, idx[:,['PRT', 'USA', 'CHN']]]
```

Com a nova implementação de `stack`, recuperamos o país como nível do índice sem eliminar implicitamente as combinações com valores em falta.

```python
serie_reposta = co2_por_ano.stack(level=[1,2])
serie_reposta.head()
```

O resultado está organizado por ano–país. Reordenamos os níveis e as linhas antes de comparar com a Series original. Repomos também o nome da Series, que não é automaticamente preservado por esta passagem por um DataFrame.

```python
serie_reposta = serie_reposta.reorder_levels(['iso_code', 'country', 'year']).sort_index()
serie_reposta.head()
```

## 9. `melt`: voltar a uma tabela de indicadores

Partimos de um subconjunto de três países e três anos para que a reorganização seja fácil de acompanhar.

```python
dados.head()
```

```python
pequeno = dados.loc[
    dados['iso_code'].isin(['PRT', 'USA', 'CHN']) & dados['year'].isin([2000, 2010, 2020]),
    ['iso_code', 'year', 'co2', 'electricity_demand_twh']
].reset_index(drop=True)
pequeno
```

Os identificadores ficam em `id_vars`; as medidas selecionadas passam a nomes de indicador e valores. As duas medidas têm unidades diferentes: a transformação não as torna comparáveis entre si.

```python
longo = pequeno.melt(
    id_vars=['iso_code', 'year'],
    value_vars=['co2', 'electricity_demand_twh'],
    var_name='indicator', value_name='value'
)
longo
```

### O que distingue as operações?
- `pivot`: reorganiza dados a partir de colunas identificadoras; exige uma única medida por combinação de chaves.
- `melt`: transforma colunas de medidas em linhas de indicador–valor.
- `unstack`: move níveis do índice das linhas para as colunas.
- `stack`: move níveis das colunas para o índice das linhas.
- `.T`: troca os dois eixos completos.
