## Exercício 6

No ficheiro `exercicios\escola.db` está uma base de dados do SQLite. Os dados estão na tabela `alunos`.

1. Leia os dados do ficheiro para um DataFrame e apresente a lista de todos os alunos do curso de Medicina.
2. Crie um DataFrame com os alunos de cada curso (um DataFrame por curso), e guarde cada um deles numa tabela da base de dados com o nome do curso.

```python
# 6.1

import pandas as pd
import sqlite3

con = sqlite3.connect(r'exercicios\escola.db')

df6_medicina = pd.read_sql('SELECT * FROM medicina ', con)

df6_medicina
```

```python
# 6.2

df6_total = pd.read_sql('SELECT Curso from Alunos', con)

df6_cursos = df6_total['Curso'].unique()

df6_cursos

for curso in df6_cursos:
    df_curso = df6_total[df6_total['Curso'] == curso]
    nome_tabela = curso.lower().replace (' ', '_')
    df_curso.to_sql(nome_tabela, con, if_exists='replace', index = False)


```

## Exercício 7

Utilizando o SQLAlchemy, conecte-se à base de dados MySQL pública `world` da [Universidade Técnica da Republica Checa](https://relational.fel.cvut.cz/)), utilizando os seguintes parâmetros:

- Host: `relational.fel.cvut.cz`
- Database: `world`
- User: `guest`
- Password: `ctu-relational`
- Port: `3306`

1. Leia tabela `Country` para um DataFrame
2. Calcule o total que população que vive nos países do sul da Europa (`Southern Europe`), e o total da área desses paises.

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

host = "relational.fel.cvut.cz"
database = "world"
user = "guest"
password = "ctu-relational"
port = 3306

url = URL.create(
    drivername="mysql+pymysql",
    username=user,
    password = password,
    host = host,
    port = port,
    database = database,
)

engine = create_engine(url)

df7 = pd.read_sql("SELECT * FROM Country", engine)

df7_soueur = df7[df7['Region'] == 'Southern Europe']

df7_soueur

total_pop = df7_soueur['Population'].sum()
total_area = df7_soueur['SurfaceArea'].sum()

print(total_area, total_pop)
```

## Exercício 8

Use o pandas para ler a página da Wikipedia com a população de cada país ([https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population](https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population))
e crie um DataFrame com essa informação, efetuando uma limpeza muito simples (removendo as linhas/colunas que não façam sentido).

```python
import pandas as pd
import requests
from io import StringIO


url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

resposta = requests.get(url, headers={"User-Agent": "Mozzila/5.0"})
tabelas8 = pd.read_html(StringIO(resposta.text))

df8 = tabelas8[0]

df8 = df8.drop(columns =['Notes','% of world', 'Date', 'Source (official or from the United Nations)'])

df8



```

## Exercicio 9

Use a API do [countries.dev](https://countries.dev/docs) para obter informações sobre países que têm o português como língua oficial.

Crie um DataFrame com as seguintes informações: nome do país, capital, população, área e idioma(s) oficiais.

```python
import requests
import pandas as pd

resposta9 = requests.get('https://countries.dev/lang/pt')

# Ver estrutura de resposta
print(resposta9.json())

# Transformar a reposta para formato json e uma variavel que vou usar para o df
dados9 = resposta9.json()

# Filtrar json pelas colunas que quero
df9 = pd.json_normalize(dados9)[['name', 'capital','population', 'area', 'languages']]

# Limpar os campos das linguas
df9['languages'] = df9['languages'].apply(lambda lista: ', '.join(l['name'] for l in lista))

df9
```
