# Dicionário das colunas

- `Employee_ID`: Employee ID (Unique number)
- `First_Name`, `Last_Name`: First and Last name of employee
- `Name`: Full name
- `Age`: Age of employee
- `Department_Region`: Compound column (e.g., "HR-Florida")
- `Status`: Employee status (Active, Inactive, Pending)
- `Join_Date`: Date entering the company
- `Salary`: Salary of employee
- `Email`: Company email
- `Phone`: Employee phone
- `Performance_Score`: Performance rating
- `Remote_Work`: Does the employee works at home?
